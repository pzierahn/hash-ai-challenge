# Create DSL search queries

````python
import json
from datetime import datetime
from pathlib import Path
from typing import Any, Literal

from IPython.display import Markdown, display
from langchain_anthropic import ChatAnthropic
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field
````

````python
openenings = Path("..") / "openings"

jobs_descriptions = [
    openenings / "ai-research-engineer-berlin.md",
    openenings / "ai-success-engineer-london.md",
    openenings / "frontend-engineer-eu-remote.md",
    openenings / "full-stack-engineer-uk-remote.md",
    openenings / "president-and-coo-london-berlin.md",
    openenings / "product-designer-ux-ui-berlin.md",
    openenings / "technical-founders-associate-berlin.md",
]

descriptions = []
for job_description_path in jobs_descriptions:
    with open(job_description_path, "r") as file:
        job_description = file.read()
    descriptions.append(job_description)
````

````python
docs = Path("..") / "docs"
multi_source_dsl_path = (
    docs / "coresignal" / "Employee APIs" / "Multi-source Employee API.json"
)

with open(multi_source_dsl_path, "r") as file:
    multi_source_dsl_json = file.read()
````

````python
system_prompt = f"""
List important fields for querying the Coresignal Multi-source Employee API. The fields should be general and applicable for all job descriptions. The API is described in the following JSON schema:

```json
{json.dumps(json.loads(multi_source_dsl_json), indent=2)}
```
""".strip()
````

````python
# llm = ChatAnthropic(model="claude-opus-5")  # type: ignore
llm = ChatAnthropic(model="claude-fable-5")  # type: ignore

# llm = ChatOpenAI(model="gpt-5.6-sol")
# dsl_gen = llm.with_structured_output(
#     DSLQuery,
#     method="json_mode",
# )
````

````python
response = llm.invoke(
    [
        ("system", system_prompt),
        (
            "human",
            "\n\n\n---\n\n\n".join(descriptions),
        ),
    ]
)
````

````python
display(Markdown(response.text))
````
